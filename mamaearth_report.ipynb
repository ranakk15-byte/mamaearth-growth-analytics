Return Analysis

```python
return_counts = orders['returned'].value_counts()

print("Return Counts:")
print(return_counts)
```

```python
total_orders = len(orders)
returned_orders = orders['returned'].sum()

return_rate = (returned_orders / total_orders) * 100

print("Total Orders:", total_orders)
print("Returned Orders:", returned_orders)
print("Return Rate:", round(return_rate, 2), "%")
```

```python
return_labels = ['Not Returned', 'Returned']
return_values = [
    (orders['returned'] == False).sum(),
    (orders['returned'] == True).sum()
]

plt.figure(figsize=(7, 5))

plt.bar(return_labels, return_values)

plt.title('Returned vs Non-Returned Orders')
plt.xlabel('Order Status')
plt.ylabel('Number of Orders')
plt.tight_layout()
plt.show()
```

```python
orders_with_city = orders.merge(
    customers[['customer_id', 'city']],
    on='customer_id',
    how='left'
)
```

```python
city_returns = (
    orders_with_city
    .groupby('city')
    .agg(
        total_orders=('order_id', 'count'),
        returned_orders=('returned', 'sum')
    )
    .reset_index()
)

city_returns['return_rate'] = (
    city_returns['returned_orders']
    / city_returns['total_orders']
    * 100
)

city_returns = city_returns.sort_values(
    'return_rate',
    ascending=False
)

city_returns
```

```python
high_return_cities = city_returns[
    city_returns['return_rate'] > 20
]

high_return_cities
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    city_returns['city'],
    city_returns['return_rate']
)

plt.title('Return Rate by City')
plt.xlabel('City')
plt.ylabel('Return Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
